```python
import os, platform, pynq
print("Board:", os.environ.get("BOARD"), "PYNQ:", pynq.__version__, "CPU:", platform.machine())
```

```python
from pynq.overlays.base import BaseOverlay
base_overlay = BaseOverlay("base.bit")
assert base_overlay.is_loaded()
help(base_overlay)
```

```python
help(base_overlay.leds)
base_overlay.leds[0].toggle()
print("LD0 toggled; verify on the board.")
```
